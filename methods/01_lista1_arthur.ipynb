```python
def get_significand(fraction: str) -> float:
    if len(fraction) != 52:
        raise Exception('Precisão inválida.')  # noqa: TRY002
    
    f = 0
    for i, b in enumerate(fraction, start = 1):
        f += int(b) * 2 ** (-i)

    return f

def get_floating_point_value(s: int, e: int, f: float) -> float:
    return (-1)**s * (1 + f) * 2 ** (e - 1023)

def get_floating_point_from_binary(bin_str: str):
    parts = bin_str.split(' ')

    if len(parts) != 3:
        raise Exception('Formato inválido.')  # noqa: TRY002

    sign, exponent, fraction = parts

    s = int(sign, 2)
    e = int(exponent, 2)
    f = get_significand(fraction)

    print(f"Sinal: {s}\nExpoente: {e}\nMantissa: {f}")

    value = get_floating_point_value(s, e, f)
    print(f"Número de ponto flutuante em decimal: {value}")
```

```python
# a) 
bin_str_a = '0 10000000101 0110100100000000000000000000000000000000000000000000'
print(f"a) {bin_str_a}")
get_floating_point_from_binary(bin_str_a)
print("\n")


# b)
bin_str_b = '1 10000000101 0110100100000000000000000000000000000000000000000000'
print(f"b) {bin_str_b}")
get_floating_point_from_binary(bin_str_b)
print("\n")

# c) 
bin_str_c = '0 10000000000 0011010100000000000000000000000000000000000000000000'
print(f"c) {bin_str_c}")
get_floating_point_from_binary(bin_str_c)
print("\n")

# d) 
bin_str_d = '0 10000000000 0011010100000000000000000000000000000000000000000001'
print(f"d) {bin_str_d}")
get_floating_point_from_binary(bin_str_d)
print("\n")
```

## Questão 3

a)
Primeiramente, vamos formular o problema como $f(t) = 0$, considerando que o objetivo é encontrar t quando T for 35ºC.

$$35 = 22 + (85 - 22)e^{-0.25t}$$

$$f(t) = 63e^{-0.25t} - 13 = 0$$

```python
import math

def f(t: float) -> float:
    return 63 * math.e ** (-0.25 * t) - 13 
```

```python
def bisection(f, a, b, tol=1e-6, max_iter=100, verbose=False):
    if a >= b or tol <= 0 or max_iter <= 0:
        return None

    fa = f(a)
    fb = f(b)

    if fa == 0:
        return a
    if fb == 0:
        return b

    if fa * fb > 0:
        return None

    p_prev = None

    for i in range(1, max_iter + 1):
        p = a + (b - a) / 2
        fp = f(p)

        if fp == 0:
            return p

        if p_prev is not None and abs(p - p_prev) < tol:
            return p

        if fa * fp > 0:
            a = p
            fa = fp
        else:
            b = p

        p_prev = p

        if verbose:
            print(f"    {str(i).zfill(4)}: {p_prev}")

    return None
```

```python
# b) Encontrar a raiz da equação usando a bisseção
interval = (2, 10) 
r = bisection(f, *interval, verbose=True)

if r is None:
    print("Não foi possível encontrar a raiz.")
else:
    print(f"O dispositivo atinge 35ºC em {r:.2f} minutos")
```

```python
def get_root_in_smaller_intervals(f, a, b, C, verbose=False) -> list:
    if a >= b or C <= 0 or C > b - a:
        return None

    _a = a

    while _a < b:
        _b = _a + C

        # Último intervalo desliga para a esquerda caso necessário
        if _b > b:
            _b = b
            _a = b - C

        fa = f(_a)
        fb = f(_b)

        if fa == 0:
            return _a
        if fb == 0:
            return _b

        if fa * fb < 0:
            if verbose:
                print(f"    Foi encontrado um intervalo de tamanho {C} válido: [{_a}, {_b}].")
                print("    Calculando pela bisseção...")
            return bisection(f, _a, _b, verbose=verbose)

        if _b == b:
            break

        _a = _b

    return None
```

```python
# c) Encontrar um intervalo válido de tamanho C e calcular a bisseção
r = get_root_in_smaller_intervals(f, 0, 100, 2, verbose=True)
if r is None:
    print("Não foi possível encontrar a raiz.")
else:
    print(f"O dispositivo atinge 35ºC em {r:.2f} minutos")
```

```python
# d) Usar as funções com o intervalo [0, 30] e C = 3
r = get_root_in_smaller_intervals(f, 0, 30, 3, verbose=True)
if r is None:
    print("Não foi possível encontrar a raiz.")
else:
    print(f"O dispositivo atinge 35ºC em {r:.2f} minutos")
```

## Questão 5

a) Seja uma função $f(x) = x^2 + x - 1 = 0$

Para a primeira função, podemos subtrair x de ambos os lados da equação:

$$x^2 + x - 1 -x = - x$$
$$x^2 - 1 = -x$$

Fatorando o polinômino, obtemos:
$$(x - 1)(x + 1) = - x$$

$$x - 1 = \frac{-x}{1 + x}$$

$$x = 1 - \frac{x}{1 + x}$$

$$x = \frac{1 + x - x}{1 + x}$$
$$x = g_1(x) = \frac{1}{1 + x}$$

O segundo caso parece apresentar um problema de sinal. Se partirmos da equação apresentada para obter a equação do segundo grau, temos:

$$x = 1 - \frac{1}{x}$$
$$x = \frac{x - 1}{x}$$
$$x^2 = x - 1$$
$$x^2 - x + 1 = 0$$

Nesse ponto, vemos que o sinal do segundo e terceiro termo estão trocados.


b) Implementação do algoritmo de iteração de Ponto Fixo

```python
def fixed_point(g, p0, tol=1e-6, max_iter=100, verbose=False):
    if tol <= 0 or max_iter <= 0:
        return None

    i = 1
    _p0 = p0

    while i <= max_iter:
        if verbose:
            print(f"    {str(i).zfill(4)}: {_p0}")

        try:
            p = g(_p0)
        except ZeroDivisionError:
            return None

        if abs(p - _p0) < tol:
            return p

        i = i + 1
        _p0 = p

    return None
```

c) Aplicação nas funções com $x_0 = 0.5$, $\epsilon = 10^{-6}$ e $N_{max} = 100$.

$$x = g_1(x) = \frac{1}{1 + x}$$

$$x = g_2(x) = 1 - \frac{1}{x}$$
 


```python
def g1(x: float) -> float:
    return 1 / (1 + x)

def g2(x: float) -> float:
    return 1 - 1 / x
```

```python
functions = (g1, g2)

for f in functions:
    resultado = fixed_point(f, p0=0.5, tol=1e-6, max_iter=100, verbose=True)

    if resultado is None:
        print(f"Função {f.__name__} - O método não convergiu no limite de iterações.\n")
    else:
        print(f"Função {f.__name__} - Aproximação do ponto fixo: {resultado}\n")
```

Observando os resultados, temos que a função g1 encontrou uma raiz em 14 iterações, enquanto a função g2 ficou em um loop e atingiu o número máximo de iterações.

Abaixo foram apresentados também resultados para a forma que acreditamos ser a correta da função $g_2(x)$. Entretanto, para o chute inicial de 0.5, ela cai num problema de divisão por zero. Se trocarmos o chute inicial para 0.4, ela consegue chegar na raiz negativa da equação, em mais iterações que a função g1.

```python

def g2_correct(x: float) -> float:
    return 1 / x - 1

functions = (g1, g2_correct)

for f in functions:
    resultado = fixed_point(f, p0=0.4, tol=1e-6, max_iter=100, verbose=True)

    if resultado is None:
        print(f"Função {f.__name__} - O método não convergiu no limite de iterações.\n")
    else:
        print(f"Função {f.__name__} - Aproximação do ponto fixo: {resultado}\n")
```

d) De acordo com os testes, podemos ver que a função g1 chegou a uma raiz em menos iterações.

Sobre a função g1, considerando o intervalo $[0, 1]$, temos os seguintes valores de bordas:

$$\frac12\leq\frac1{1+x}\leq1.$$

Ou seja, $g(x) \subseteq[0,1]$ quando  $x\in[0,1]$. Quanto à derivada de g1, temos:


$$|g_1'(x)| = \frac{1}{(1+x)^2}$$


É possível perceber que para $x = 0$, obtemos $|g_1'(x)| = 1$, o que viola as condições do teorema, pois quanto mais x de aproxima de 0, mais perto a derivada fica de 1, o que não é uma constante fixa k. Entretanto, ao tomar a próxima iteração do intervalo, sendo:

$$x_{n + 1} = \frac{1}{1 + x_n}$$

Se $x_n \in [0, 1]$ então, $1 + x_n \in [1, 2]$. Sendo assim:

$$\frac{1}{1 + x_n} \in [0.5, 1]$$

Então, na próxima iteração, o intervalo se torna $[0.5, 1]$. Nesse caso, temos que x é sempre maior ou igual a 0.5, então:

$$x \geq \frac12$$

$$(1 + x)^2 \geq (1 + \frac12)^2$$
$$(1 + x)^2 \geq \frac94$$

$$|g_1'(x)| = \frac{1}{(1+x)^2} \leq \frac49 < 1$$

O que satisfaz o teorema.

Já a formulação corrigida $(g_2(x)=\frac1x-1)$ não está definida em zero e não leva todo o intervalo $([0,1])$ para si mesmo. Ela não satisfaz as hipóteses do teorema nesse intervalo.

Assim, $(g_1)$ é mais adequada, pois sua convergência é garantida para qualquer aproximação inicial em $([0,1])$.
