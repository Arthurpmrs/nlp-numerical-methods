```python
def bissection(f, a, b, tol, n_steps):
    if f(a) * f(b) > 0:
        raise Exception(f"The function does not have a root in the interval [{a}, {b}]")
    
    i = 0
    fm = f(a)
    a_it = a
    b_it = b
    root_convergence_seq = []

    while i < n_steps:
        m = (a_it + b_it) / 2
        fa = f(a_it)
        fb = f(b_it)  
        fm = f(m)
        root_convergence_seq.append(m)
        print(a_it, b_it, m)

        if fm == 0 or (b - a) / 2 < tol:
            print('success')
            break
        
        if fa * fm > 0:
            a_it = m
        else:
            b_it = m
        
        i += 1

    


    print(root_convergence_seq)
    print(root_convergence_seq[-1])
```

```python
def polynomial(x):
    return x ** 3 + 4 * x ** 2 - 10         
```
