```python
import pandas as pd
```

```python
original = pd.read_csv('data/songs_reduced.csv')
```

```python
print(original)
```

```python
import re


def preprocess_lyrics(text):
    if pd.isna(text):
        return ""

    text = text.lower()
    text = re.sub(r"\b\w*\d\w*\b", "", text)
    text = re.sub(r"[^\w\s]", "", text)
    text = re.sub(r"\s+", " ", text).strip()

    return text
```

```python
original['lyrics'] = original['lyrics'].apply(preprocess_lyrics)
```

```python
original["tokens"] = original["lyrics"].str.split()
```

```python
original[['genre', 'lyrics', 'tokens']]
```

```python
vocab = set()

for tokens in original['tokens']:
    vocab.update(tokens)

vocab
```
