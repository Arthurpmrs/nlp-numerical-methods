# Filtragem da base original
A [base original](https://www.kaggle.com/datasets/serkantysz/550k-spotify-songs-audio-lyrics-and-genres) consiste em uma lista de músicas do Spotify com metadados e letra. Quanto ao gênero, a base original possui 10 gêneros principais, com quantidades que variam.

Como o objetivo é avaliar a letra, foi feita uma filtragem que deixa apenas o id, nome da música, artistas, letra e gênero. Foram também removidos gêneros com pouca ou nenhuma letra, deixando apenas Rock, Pop, Folk, Country e Hip-Hop. Quanto à quantidade, foram selecionados de foram aleatória `n` músicas de cada um dos gêneros selecionados, para manter a base uniforme.

```python
import pandas as pd

GENRE_SAMPLE_SIZE = 4000
```

```python
# Download e extração da base de dados
!curl -L -o ./data/550k-spotify-songs-audio-lyrics-and-genres.zip\
  https://www.kaggle.com/api/v1/datasets/download/serkantysz/550k-spotify-songs-audio-lyrics-and-genres

!unzip -o ./data/550k-spotify-songs-audio-lyrics-and-genres.zip -d data/
```

```python
original = pd.read_csv('data/songs.csv')
```

```python
# Colunas da base original
original.columns
```

```python
filtered_columns = original[['id', 'name', 'artists', 'lyrics', 'genre']]

# Quantidade de instâncias por gênero na base original
filtered_columns['genre'].value_counts()
```

```python
# Fazer sample da base original para uniformizar e reduzir tamanho 
selected_genres = ["Rock", "Pop", "Folk", "Country", "Hip-Hop"]

filtered_data = (
    filtered_columns[filtered_columns["genre"].isin(selected_genres)]
    .groupby("genre", group_keys=False)
    .sample(n=GENRE_SAMPLE_SIZE, random_state=42)
    .sample(frac=1, random_state=42)
    .reset_index(drop=True)
)
filtered_data['genre'].value_counts()
```

```python
# Exportar para arquivo
filtered_data.to_csv("data/songs_reduced.csv", index=False)
```
